```python
import getpass
import base64
from langchain_openai import ChatOpenAI    # ← replaces google.generativeai

# 🔑 Securely input your AICredits API key
API_KEY = getpass.getpass("Enter your AICredits API key: ")
API_KEY = API_KEY.strip()

# Convert image to base64 function
def image_to_base64(image_path):
    ext = image_path.split(".")[-1].lower()
    format_map = {
        "jpg": "image/jpeg",
        "jpeg": "image/jpeg",
        "png": "image/png",
        "gif": "image/gif",
        "webp": "image/webp"
    }
    mime_type = format_map.get(ext, "image/jpeg")
    with open(image_path, "rb") as f:
        encoded = base64.b64encode(f.read()).decode("utf-8")
    return encoded, mime_type

# Load both images — same as before ✅
img1_data, img1_mime = image_to_base64("cat.jpg")
img2_data, img2_mime = image_to_base64("273.jpg")

# Initialize Claude model
model = ChatOpenAI(
    model="claude-haiku-4-5",             # same as gemini-2.5-flash
    api_key=API_KEY,
    base_url="https://api.aicredits.in/v1"
)

# Build prompt with BOTH images — same as [prompt, img1, img2]
prompt = [
    {
        "role": "user",
        "content": [
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:{img1_mime};base64,{img1_data}"  # image 1
                }
            },
            {
                "type": "image_url",
                "image_url": {
                    "url": f"data:{img2_mime};base64,{img2_data}"  # image 2
                }
            },
            {
                "type": "text",
                "text": "compare both image you see"  # same prompt ✅
            }
        ]
    }
]

# Get response
response = model.invoke(prompt)

# 💬 Print answer
print("🤖 Claude says:", response.content)  # ← same as response.text
```

```python
from getpass import getpass
from openai import OpenAI

api_key = getpass("Enter API Key: ")

client = OpenAI(
    api_key=api_key,
    base_url="https://api.aicredits.in/v1"
)

messages = [
    {
        "role": "system",
        "content": "You are a helpful assistant."
    }
]

while True:

    question = input("\nYou: ")

    if question.lower() == "exit":
        print("Goodbye!")
        break

    messages.append(
        {
            "role": "user",
            "content": question
        }
    )

    response = client.chat.completions.create(
        model="anthropic/claude-sonnet-4.6",
        messages=messages
    )

    answer = response.choices[0].message.content

    print("\nClaude:", answer)

    messages.append(
        {
            "role": "assistant",
            "content": answer
        }
    )
```
